```python
import torch
import gc

# Clear memory
gc.collect()
torch.cuda.empty_cache()
```

```python
from faster_whisper import WhisperModel
import os
import csv
from tqdm import tqdm

# --- 1. CONFIG ---
model_size = "small"
model = WhisperModel(model_size, device="cuda", compute_type="float16")

audio_dir = r"/content/audio_data/Audio"
csv_file = "transcriptions_v2.csv"

audio_files = [f for f in os.listdir(audio_dir) if f.endswith((".mp3", ".wav"))]

# --- 2. LIVE PROCESSING ---
print(f"Total Files: {len(audio_files)}")

# 'a' (append)
with open(csv_file, 'w', newline='', encoding='utf-8') as f:
    writer = csv.writer(f)
    writer.writerow(["Audio Link", "Transcription"])


    for filename in tqdm(audio_files, desc="Processing Urdu Audio"):
        path = os.path.join(audio_dir, filename)
        try:
            segments, _ = model.transcribe(path, beam_size=1, language="ur")
            text = "".join([s.text for s in segments]).strip()


            writer.writerow([filename, text])
            f.flush() # Force save to disk

        except Exception as e:
            print(f"\nError in {filename}: {e}")

print(f"\nDone! Check File: {csv_file}")
```
