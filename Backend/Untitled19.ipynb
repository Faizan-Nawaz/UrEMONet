```python
import os
import whisper
import csv

# --- 1. SETTINGS ---
# Initialize the Whisper model using the 'large' configuration 
model = whisper.load_model('large')

# Set your directory using the raw string to avoid syntax errors 
audio_directory = r"C:\Users\MALIIK\Desktop\new_audio_data"
csv_file = "transcriptions.csv"
transcription_data = []

# --- 2. PROCESSING ---
# Iterate through each file in your folder 
for filename in os.listdir(audio_directory):
    if filename.endswith(".mp3") or filename.endswith(".wav"):
        audio_path = os.path.join(audio_directory, filename)
        
        print(f"Processing: {filename}")
        try:
            # Perform Urdu speech recognition 
            result = model.transcribe(audio_path)
            
            # Store the path and its transcription 
            transcription_data.append([audio_path, result['text']])
            print(f"Done: {result['text'][:30]}...")
            
        except Exception as e:
            print(f"Error processing {filename}: {str(e)}")

# --- 3. OUTPUT ---
# Save the results to the CSV for integration with other modalities 
with open(csv_file, 'w', newline='', encoding='utf-8') as f:
    writer = csv.writer(f)
    writer.writerow(["Audio Link", "Transcription"])
    writer.writerows(transcription_data)

print(f"\nTranscription complete. File saved to: {csv_file}")
```
