**Extracting Audios from Video Dataset**

```python
import os
from moviepy.editor import VideoFileClip
from tqdm import tqdm

# --- CONFIGURATION ---
video_directory = '/content/audio_data/video'
output_directory = '/content/extracted_audio'
os.makedirs(output_directory, exist_ok=True)

# Get list of videos
video_files = [f for f in os.listdir(video_directory) if f.endswith(".mp4")]

print(f"Total videos found: {len(video_files)}")

# --- EXTRACTION LOOP ---
for video_file in tqdm(video_files, desc="Extracting Audio"):
    video_path = os.path.join(video_directory, video_file)
    audio_filename = f"{os.path.splitext(video_file)[0]}.mp3"
    output_audio_path = os.path.join(output_directory, audio_filename)

    # Skip if already extracted (Resume Logic)
    if os.path.exists(output_audio_path):
        continue

    try:
        video_clip = VideoFileClip(video_path)

        # Check if video actually has audio
        if video_clip.audio is not None:
            # logger=None hides the messy moviepy console output
            video_clip.audio.write_audiofile(output_audio_path, logger=None)
            video_clip.audio.close()

        video_clip.close()
    except Exception as e:
        print(f"Error processing {video_file}: {e}")

print(f"\nDone! Audio files are saved in: {output_directory}")
```
